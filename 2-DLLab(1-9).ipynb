```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

def step_function(x):
    return 1 if x >= 0 else 0
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
def sigmoid_derivative(x):
    return x * (1 - x)
print("\n--- PART A: AND Gate with Single Neuron ---")
X_and = np.array([[0,0],[0,1],[1,0],[1,1]])
Y_and = np.array([0,0,0,1])
w = np.random.rand(2)
b = np.random.rand(1)
lr = 0.1
losses_and = []

for epoch in range(15):
    total_error = 0
    for i in range(len(X_and)):
        linear_output = np.dot(X_and[i], w) + b
        y_pred = step_function(linear_output)
        error = Y_and[i] - y_pred
        w += lr * error * X_and[i]
        b += lr * error
        total_error += abs(error)
    losses_and.append(total_error)

print("Final Weights:", w, "Bias:", b)
print("\nTesting AND Gate:")
for x in X_and:
    print(f"{x} -> {step_function(np.dot(x, w) + b)}")
```

```python
print("\n--- PART B: XOR using Feedforward NN ---")
X_xor = np.array([[0,0],[0,1],[1,0],[1,1]])
Y_xor = np.array([[0],[1],[1],[0]])
np.random.seed(42)
wh = np.random.uniform(size=(2, 2))
bh = np.random.uniform(size=(1, 2))
wout = np.random.uniform(size=(2, 1))
bout = np.random.uniform(size=(1, 1))
lr = 0.5
epochs = 10000
losses_xor = []
for epoch in range(epochs):
    hidden_input = np.dot(X_xor, wh) + bh
    hidden_output = sigmoid(hidden_input)
    final_input = np.dot(hidden_output, wout) + bout
    final_output = sigmoid(final_input)
    error = Y_xor - final_output
    losses_xor.append(np.mean(np.square(error)))
    d_output = error * sigmoid_derivative(final_output)
    error_hidden = d_output.dot(wout.T)
    d_hidden = error_hidden * sigmoid_derivative(hidden_output)
    wout += hidden_output.T.dot(d_output) * lr
    bout += np.sum(d_output, axis=0, keepdims=True) * lr
    wh += X_xor.T.dot(d_hidden) * lr
    bh += np.sum(d_hidden, axis=0, keepdims=True) * lr
print("\nTesting XOR Gate:")
for x in X_xor:
    hidden = sigmoid(np.dot(x, wh) + bh)
    output = sigmoid(np.dot(hidden, wout) + bout)
    print(f"{x} -> {np.round(output[0], 2)}")
```

```python
print("\n--- PART C: MLP on make_moons Dataset ---")
X, Y = make_moons(n_samples=500, noise=0.2, random_state=42)
Y = Y.reshape(-1, 1)
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
wh = np.random.uniform(size=(2, 4))
bh = np.zeros((1, 4))
wout = np.random.uniform(size=(4, 1))
bout = np.zeros((1, 1))
lr = 0.1
epochs = 10000
losses_mlp = []
for epoch in range(epochs):
    hidden_input = np.dot(X_train, wh) + bh
    hidden_output = sigmoid(hidden_input)
    final_input = np.dot(hidden_output, wout) + bout
    final_output = sigmoid(final_input)
    error = Y_train - final_output
    losses_mlp.append(np.mean(np.square(error)))
    d_output = error * sigmoid_derivative(final_output)
    error_hidden = d_output.dot(wout.T)
    d_hidden = error_hidden * sigmoid_derivative(hidden_output)
    wout += hidden_output.T.dot(d_output) * lr
    bout += np.sum(d_output, axis=0, keepdims=True) * lr
    wh += X_train.T.dot(d_hidden) * lr
    bh += np.sum(d_hidden, axis=0, keepdims=True) * lr
hidden = sigmoid(np.dot(X_test, wh) + bh)
output = sigmoid(np.dot(hidden, wout) + bout)
predictions = (output > 0.5).astype(int)
accuracy = np.mean(predictions == Y_test) * 100
print(f"MLP Accuracy: {accuracy:.2f}%")
```
