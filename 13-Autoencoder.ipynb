```python
# Part I: Convolutional Autoencoder on MNIST
# Save as part1_autoencoder.py or run in a notebook cell.

import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np
import matplotlib.pyplot as plt

# Load MNIST
(x_train, _), (x_test, _) = tf.keras.datasets.mnist.load_data()
x_train = x_train.astype("float32") / 255.0
x_test  = x_test.astype("float32")  / 255.0
x_train = np.expand_dims(x_train, -1)  # (N,28,28,1)
x_test  = np.expand_dims(x_test, -1)

# Model parameters
input_shape = (28,28,1)
latent_dim = 64

# Encoder
encoder_inputs = layers.Input(shape=input_shape)
x = layers.Conv2D(32, 3, activation="relu", padding="same", strides=2)(encoder_inputs)  # 14x14x32
x = layers.Conv2D(64, 3, activation="relu", padding="same", strides=2)(x)              # 7x7x64
x = layers.Flatten()(x)
z = layers.Dense(latent_dim, name="latent")(x)
encoder = models.Model(encoder_inputs, z, name="encoder")
encoder.summary()

# Decoder
latent_inputs = layers.Input(shape=(latent_dim,))
x = layers.Dense(7*7*64, activation="relu")(latent_inputs)
x = layers.Reshape((7,7,64))(x)
x = layers.Conv2DTranspose(64, 3, strides=2, padding="same", activation="relu")(x)  # 14x14
x = layers.Conv2DTranspose(32, 3, strides=2, padding="same", activation="relu")(x)  # 28x28
decoder_outputs = layers.Conv2D(1, 3, padding="same", activation="sigmoid")(x)
decoder = models.Model(latent_inputs, decoder_outputs, name="decoder")
decoder.summary()

# Autoencoder
ae_inputs = encoder_inputs
ae_latent = encoder(ae_inputs)
ae_outputs = decoder(ae_latent)
autoencoder = models.Model(ae_inputs, ae_outputs, name="autoencoder")
autoencoder.compile(optimizer='adam', loss='binary_crossentropy')
autoencoder.summary()

# Train
autoencoder.fit(
    x_train, x_train,
    epochs=20,
    batch_size=128,
    shuffle=True,
    validation_data=(x_test, x_test)
)

# Visualize some reconstructions
n = 10
idx = np.random.randint(0, x_test.shape[0], n)
recon = autoencoder.predict(x_test[idx])
plt.figure(figsize=(20,4))
for i in range(n):
    # original
    ax = plt.subplot(2,n,i+1)
    plt.imshow(x_test[idx[i]].squeeze(), cmap='gray')
    plt.title("orig")
    plt.axis('off')
    # reconstructed
    ax = plt.subplot(2,n,i+1+n)
    plt.imshow(recon[i].squeeze(), cmap='gray')
    plt.title("recon")
    plt.axis('off')
plt.show()
```

```python
# Part III: Simple DCGAN on MNIST
# Save as part3_gan.py or run in a notebook cell.

import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np
import matplotlib.pyplot as plt
import os

# Load MNIST
(x_train, _), (_, _) = tf.keras.datasets.mnist.load_data()
x_train = (x_train.astype('float32') - 127.5) / 127.5  # scale to [-1,1]
x_train = np.expand_dims(x_train, axis=-1)
BUFFER_SIZE = x_train.shape[0]
BATCH_SIZE = 256
dataset = tf.data.Dataset.from_tensor_slices(x_train).shuffle(BUFFER_SIZE).batch(BATCH_SIZE)

# Model parameters
noise_dim = 100
epochs = 50
sample_dir = 'gan_samples'
os.makedirs(sample_dir, exist_ok=True)

# Generator
def make_generator_model():
    model = tf.keras.Sequential()
    model.add(layers.Dense(7*7*256, use_bias=False, input_shape=(noise_dim,)))
    model.add(layers.BatchNormalization())
    model.add(layers.LeakyReLU())
    model.add(layers.Reshape((7,7,256)))
    model.add(layers.Conv2DTranspose(128, 5, strides=1, padding='same', use_bias=False))
    model.add(layers.BatchNormalization())
    model.add(layers.LeakyReLU())
    model.add(layers.Conv2DTranspose(64, 5, strides=2, padding='same', use_bias=False))
    model.add(layers.BatchNormalization())
    model.add(layers.LeakyReLU())
    model.add(layers.Conv2DTranspose(1, 5, strides=2, padding='same', use_bias=False, activation='tanh'))
    return model

# Discriminator
def make_discriminator_model():
    model = tf.keras.Sequential()
    model.add(layers.Conv2D(64, 5, strides=2, padding='same', input_shape=[28,28,1]))
    model.add(layers.LeakyReLU())
    model.add(layers.Dropout(0.3))
    model.add(layers.Conv2D(128, 5, strides=2, padding='same'))
    model.add(layers.LeakyReLU())
    model.add(layers.Dropout(0.3))
    model.add(layers.Flatten())
    model.add(layers.Dense(1))
    return model

generator = make_generator_model()
discriminator = make_discriminator_model()

# Losses and optimizers
cross_entropy = tf.keras.losses.BinaryCrossentropy(from_logits=True)

def discriminator_loss(real_output, fake_output):
    real_loss = cross_entropy(tf.ones_like(real_output), real_output)
    fake_loss = cross_entropy(tf.zeros_like(fake_output), fake_output)
    return real_loss + fake_loss

def generator_loss(fake_output):
    return cross_entropy(tf.ones_like(fake_output), fake_output)

gen_optimizer = tf.keras.optimizers.Adam(1e-4)
disc_optimizer = tf.keras.optimizers.Adam(1e-4)

# Training step
@tf.function
def train_step(images):
    noise = tf.random.normal([BATCH_SIZE, noise_dim])

    with tf.GradientTape() as gen_tape, tf.GradientTape() as disc_tape:
        generated_images = generator(noise, training=True)

        real_output = discriminator(images, training=True)
        fake_output = discriminator(generated_images, training=True)

        gen_loss = generator_loss(fake_output)
        disc_loss = discriminator_loss(real_output, fake_output)

    gradients_of_generator = gen_tape.gradient(gen_loss, generator.trainable_variables)
    gradients_of_discriminator = disc_tape.gradient(disc_loss, discriminator.trainable_variables)

    gen_optimizer.apply_gradients(zip(gradients_of_generator, generator.trainable_variables))
    disc_optimizer.apply_gradients(zip(gradients_of_discriminator, discriminator.trainable_variables))

    return gen_loss, disc_loss

# Training loop
def train(dataset, epochs):
    for epoch in range(epochs):
        gen_loss_avg = 0.0
        disc_loss_avg = 0.0
        batches = 0
        for image_batch in dataset:
            # ensure batch has correct size for @tf.function (drop last small batch)
            if image_batch.shape[0] != BATCH_SIZE:
                continue
            g_loss, d_loss = train_step(image_batch)
            gen_loss_avg += g_loss
            disc_loss_avg += d_loss
            batches += 1
        if batches > 0:
            gen_loss_avg /= batches
            disc_loss_avg /= batches
        print(f"Epoch {epoch+1}/{epochs} — gen_loss: {gen_loss_avg:.4f} disc_loss: {disc_loss_avg:.4f}")

        # produce sample
        display.clear_output(wait=True)
        generate_and_save_images(generator, epoch+1, tf.random.normal([16, noise_dim]))

def generate_and_save_images(model, epoch, test_input):
    predictions = model(test_input, training=False)
    fig = plt.figure(figsize=(4,4))
    for i in range(predictions.shape[0]):
        plt.subplot(4,4,i+1)
        plt.imshow((predictions[i,:,:,0] * 127.5 + 127.5).numpy().astype(np.uint8), cmap='gray')
        plt.axis('off')
    plt.suptitle(f"Epoch {epoch}")
    plt.tight_layout()
    plt.savefig(os.path.join(sample_dir, f'image_at_epoch_{epoch:03d}.png'))
    plt.close(fig)

# Imports for display usage in training (not mandatory)
from IPython import display

# Run training (this may take a while depending on your hardware)
train(dataset, epochs=epochs)

# Generate final samples
seed = tf.random.normal([16, noise_dim])
generate_and_save_images(generator, epochs, seed)
```

```python
"""
Separated experiments (AE, VAE, DCGAN) with evaluation metrics and curves.
Save results to OUTPUT_DIR/AE, OUTPUT_DIR/VAE, OUTPUT_DIR/GAN respectively.
Requirements: pip install torch torchvision matplotlib scikit-image tqdm

Usage examples:
    python exp12-14_pytorch_separated.py --run ae
    python exp12-14_pytorch_separated.py --run vae
    python exp12-14_pytorch_separated.py --run gan
    python exp12-14_pytorch_separated.py         # runs all sequentially

The file consolidates and improves the script you provided: clearer modular functions,
consistent saving, per-experiment metrics JSON, and additional evaluation curves/plots.
"""

import os, time, json, math, random, argparse
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader
from torchvision import transforms, datasets, utils
from skimage.metrics import structural_similarity as ssim

# ----------------- Config -----------------
OUTPUT_DIR = Path("./exp_results_pytorch_separated")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
SEED = 42
torch.manual_seed(SEED); np.random.seed(SEED); random.seed(SEED)

# default hyperparams
BATCH = 128
IMG_SIZE = 28
IMG_CHANNELS = 1
EPOCHS_AE = 10
EPOCHS_VAE = 10
GAN_STEPS = 2000
LATENT_AE = 64
LATENT_VAE = 2
LATENT_GAN = 100

# ---------- utils ----------

def psnr_np(a, b):
    mse = np.mean((a - b) ** 2)
    if mse == 0:
        return 999.0
    PIXEL_MAX = 1.0
    return 20 * math.log10(PIXEL_MAX / math.sqrt(mse))


def save_fig(fig, path):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    fig.tight_layout()
    fig.savefig(path, dpi=150, bbox_inches='tight')
    plt.close(fig)
    print(f"Saved: {path}")


def to_np(tensor):
    return tensor.detach().cpu().numpy()


# --------------- Data loader ----------------

def get_dataloaders(batch=BATCH):
    transform = transforms.Compose([transforms.ToTensor()])
    train_dataset = datasets.MNIST(root="./data", train=True, download=True, transform=transform)
    test_dataset  = datasets.MNIST(root="./data", train=False, download=True, transform=transform)
    train_loader = DataLoader(train_dataset, batch_size=batch, shuffle=True, num_workers=2, pin_memory=True)
    test_loader  = DataLoader(test_dataset, batch_size=batch, shuffle=False, num_workers=2, pin_memory=True)
    return train_loader, test_loader, train_dataset, test_dataset


# --------------- Models (same as provided, cleaned) ----------------
class ConvAutoencoder(nn.Module):
    def __init__(self, latent_dim=64):
        super().__init__()
        self.enc = nn.Sequential(
            nn.Conv2d(1,32,3,padding=1), nn.ReLU(),
            nn.MaxPool2d(2,ceil_mode=True),
            nn.Conv2d(32,64,3,padding=1), nn.ReLU(),
            nn.MaxPool2d(2,ceil_mode=True),
        )
        self.flatten = nn.Flatten()
        self.fc_enc = nn.Linear(7*7*64, latent_dim)
        self.fc_dec = nn.Linear(latent_dim, 7*7*64)
        self.dec = nn.Sequential(
            nn.Upsample(scale_factor=2, mode='nearest'),
            nn.ConvTranspose2d(64,64,3,padding=1), nn.ReLU(),
            nn.Upsample(scale_factor=2, mode='nearest'),
            nn.ConvTranspose2d(64,32,3,padding=1), nn.ReLU(),
            nn.Conv2d(32,1,3,padding=1), nn.Sigmoid()
        )
    def forward(self, x):
        x_e = self.enc(x)
        x_f = self.flatten(x_e)
        z = self.fc_enc(x_f)
        x_d = self.fc_dec(z).view(-1,64,7,7)
        out = self.dec(x_d)
        return out, z


class VAE(nn.Module):
    def __init__(self, latent_dim=2):
        super().__init__()
        self.enc_conv = nn.Sequential(
            nn.Conv2d(1,32,3,padding=1), nn.ReLU(),
            nn.MaxPool2d(2,ceil_mode=True),
            nn.Conv2d(32,64,3,padding=1), nn.ReLU(),
            nn.MaxPool2d(2,ceil_mode=True),
            nn.Flatten()
        )
        self.fc1 = nn.Linear(7*7*64, 128)
        self.fc_mu = nn.Linear(128, latent_dim)
        self.fc_logvar = nn.Linear(128, latent_dim)
        self.fc_decode = nn.Linear(latent_dim, 7*7*64)
        self.decoder = nn.Sequential(
            nn.Upsample(scale_factor=2, mode='nearest'),
            nn.ConvTranspose2d(64,64,3,padding=1), nn.ReLU(),
            nn.Upsample(scale_factor=2, mode='nearest'),
            nn.ConvTranspose2d(64,32,3,padding=1), nn.ReLU(),
            nn.Conv2d(32,1,3,padding=1), nn.Sigmoid()
        )
    def encode(self, x):
        h = self.enc_conv(x)
        h = F.relu(self.fc1(h))
        mu = self.fc_mu(h)
        logvar = self.fc_logvar(h)
        return mu, logvar
    def reparameterize(self, mu, logvar):
        std = torch.exp(0.5*logvar)
        eps = torch.randn_like(std)
        return mu + eps * std
    def decode(self, z):
        h = F.relu(self.fc_decode(z))
        h = h.view(-1,64,7,7)
        return self.decoder(h)
    def forward(self, x):
        mu, logvar = self.encode(x)
        z = self.reparameterize(mu, logvar)
        x_recon = self.decode(z)
        return x_recon, mu, logvar, z


class DCGAN_G(nn.Module):
    def __init__(self, nz=100):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(nz, 7*7*256, bias=False),
            nn.BatchNorm1d(7*7*256), nn.LeakyReLU(0.2, inplace=True)
        )
        self.deconv = nn.Sequential(
            nn.ConvTranspose2d(256,128,kernel_size=5,stride=1,padding=2,bias=False),
            nn.BatchNorm2d(128), nn.LeakyReLU(0.2, inplace=True),
            nn.ConvTranspose2d(128,64,kernel_size=5,stride=2,padding=2,output_padding=1,bias=False),
            nn.BatchNorm2d(64), nn.LeakyReLU(0.2, inplace=True),
            nn.ConvTranspose2d(64,1,kernel_size=5,stride=2,padding=2,output_padding=1,bias=False),
            nn.Tanh()
        )
    def forward(self, z):
        x = self.net(z).view(-1,256,7,7)
        x = self.deconv(x)
        return x

class DCGAN_D(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv2d(1,64,kernel_size=5,stride=2,padding=2),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Dropout(0.3),
            nn.Conv2d(64,128,kernel_size=5,stride=2,padding=2),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Dropout(0.3),
            nn.Flatten(),
            nn.Linear(7*7*128,1)
        )
    def forward(self, x):
        return self.net(x)


# ---------------- AE experiment ----------------
def run_ae(output_dir=OUTPUT_DIR/"AE", epochs=EPOCHS_AE, latent_dim=LATENT_AE, batch=BATCH):
    output_dir = Path(output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)
    train_loader, test_loader, _, _ = get_dataloaders(batch=batch)

    ae = ConvAutoencoder(latent_dim=latent_dim).to(DEVICE)
    ae_opt = torch.optim.Adam(ae.parameters(), lr=1e-3)
    criterion_mse = nn.MSELoss()

    train_losses, val_losses = [], []
    t0 = time.time()
    for epoch in range(epochs):
        ae.train()
        train_loss = 0.0
        for xb, _ in train_loader:
            xb = xb.to(DEVICE)
            out, _ = ae(xb)
            loss = criterion_mse(out, xb)
            ae_opt.zero_grad(); loss.backward(); ae_opt.step()
            train_loss += loss.item() * xb.size(0)
        train_loss /= len(train_loader.dataset)
        train_losses.append(train_loss)
        # val
        ae.eval()
        val_loss = 0.0
        with torch.no_grad():
            for xb, _ in test_loader:
                xb = xb.to(DEVICE)
                out, _ = ae(xb)
                loss = criterion_mse(out, xb)
                val_loss += loss.item() * xb.size(0)
        val_loss /= len(test_loader.dataset)
        val_losses.append(val_loss)
        print(f"AE Epoch {epoch+1}/{epochs} TrainLoss {train_loss:.6f} ValLoss {val_loss:.6f}")
    ae_time = time.time() - t0

    # collect reconstructions
    ae.eval()
    recons_list = []; orig_list = []
    with torch.no_grad():
        for xb, _ in test_loader:
            xb = xb.to(DEVICE)
            out, _ = ae(xb)
            recons_list.append(out.cpu()); orig_list.append(xb.cpu())
    recons_all = torch.cat(recons_list, dim=0).numpy()
    orig_all = torch.cat(orig_list, dim=0).numpy()

    mses = np.mean((orig_all - recons_all)**2, axis=(1,2,3))
    psnrs = [psnr_np(orig_all[i].squeeze(), recons_all[i].squeeze()) for i in range(len(orig_all))]
    ssims = [ssim(orig_all[i].squeeze(), recons_all[i].squeeze(), data_range=1.0) for i in range(len(orig_all))]

    metrics = {
        "mse_mean": float(np.mean(mses)), "mse_std": float(np.std(mses)),
        "psnr_mean": float(np.mean(psnrs)), "psnr_std": float(np.std(psnrs)),
        "ssim_mean": float(np.mean(ssims)), "ssim_std": float(np.std(ssims)),
        "train_time_sec": ae_time
    }
    print("AE metrics:", metrics)

    # save metrics
    (output_dir/"metrics.json").write_text(json.dumps(metrics, indent=2))

    # loss curve
    fig = plt.figure(figsize=(6,4))
    plt.plot(train_losses); plt.plot(val_losses)
    plt.title('AE Loss (MSE)'); plt.xlabel('Epoch'); plt.ylabel('Loss'); plt.legend(['train','val'])
    save_fig(fig, output_dir/"ae_loss.png")

    # recon samples
    n = 10
    idx = np.random.choice(len(orig_all), n, replace=False)
    fig = plt.figure(figsize=(20,4))
    for i, k in enumerate(idx):
        ax=plt.subplot(2,n,i+1); plt.imshow(orig_all[k].squeeze(), cmap='gray'); plt.axis('off')
        ax=plt.subplot(2,n,n+i+1); plt.imshow(recons_all[k].squeeze(), cmap='gray'); plt.axis('off')
    plt.suptitle('AE: top original, bottom reconstruction', y=1.02)
    save_fig(fig, output_dir/"ae_recon.png")

    # histogram of PSNR and SSIM
    fig = plt.figure(figsize=(10,4))
    plt.subplot(1,2,1); plt.hist(psnrs, bins=50); plt.title('PSNR distribution'); plt.xlabel('PSNR')
    plt.subplot(1,2,2); plt.hist(ssims, bins=50); plt.title('SSIM distribution'); plt.xlabel('SSIM')
    save_fig(fig, output_dir/"ae_metrics_hist.png")

    return metrics


# ---------------- VAE experiment ----------------
def vae_loss_function(recon_x, x, mu, logvar):
    BCE = F.binary_cross_entropy(recon_x, x, reduction='sum')
    KLD = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp())
    return (BCE + KLD) / x.size(0)


def run_vae(output_dir=OUTPUT_DIR/"VAE", epochs=EPOCHS_VAE, latent_dim=LATENT_VAE, batch=BATCH):
    output_dir = Path(output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)
    train_loader, test_loader, _, test_dataset = get_dataloaders(batch=batch)

    vae = VAE(latent_dim=latent_dim).to(DEVICE)
    opt = torch.optim.Adam(vae.parameters(), lr=1e-3)
    train_losses = []
    t0 = time.time()
    for epoch in range(epochs):
        vae.train()
        train_loss = 0.0
        for xb, _ in train_loader:
            xb = xb.to(DEVICE)
            recon, mu, logvar, _ = vae(xb)
            loss = vae_loss_function(recon, xb, mu, logvar)
            opt.zero_grad(); loss.backward(); opt.step()
            train_loss += loss.item() * xb.size(0)
        train_loss /= len(train_loader.dataset)
        train_losses.append(train_loss)
        print(f"VAE Epoch {epoch+1}/{epochs} TrainLoss {train_loss:.6f}")
    vae_time = time.time() - t0

    # eval
    vae.eval()
    recons_list=[]; mus_list=[]; zs_list=[]
    with torch.no_grad():
        for xb, _ in test_loader:
            xb = xb.to(DEVICE)
            recon, mu, logvar, z = vae(xb)
            recons_list.append(recon.cpu()); mus_list.append(mu.cpu()); zs_list.append(z.cpu())
    recons_all = torch.cat(recons_list, dim=0).numpy()
    mus_all = torch.cat(mus_list, dim=0).numpy()

    # orig (reuse test dataset to compute metrics)
    orig = np.array(test_dataset.data.unsqueeze(1).float()/255.0)

    mses = np.mean((orig - recons_all)**2, axis=(1,2,3))
    psnrs = [psnr_np(orig[i].squeeze(), recons_all[i].squeeze()) for i in range(len(orig))]
    ssims = [ssim(orig[i].squeeze(), recons_all[i].squeeze(), data_range=1.0) for i in range(len(orig))]

    metrics = {
        "mse_mean": float(np.mean(mses)), "mse_std": float(np.std(mses)),
        "psnr_mean": float(np.mean(psnrs)), "psnr_std": float(np.std(psnrs)),
        "ssim_mean": float(np.mean(ssims)), "ssim_std": float(np.std(ssims)),
        "train_time_sec": vae_time
    }
    print("VAE metrics:", metrics)
    (output_dir/"metrics.json").write_text(json.dumps(metrics, indent=2))

    # loss curve
    fig = plt.figure(figsize=(6,4))
    plt.plot(train_losses); plt.title('VAE Loss (recon + KL)'); plt.xlabel('Epoch'); plt.ylabel('Loss')
    save_fig(fig, output_dir/"vae_loss.png")

    # latent scatter (if 2D)
    if latent_dim == 2:
        test_labels = np.concatenate([y.numpy() for (_, y) in DataLoader(test_dataset, batch_size=BATCH)], axis=0)
        fig = plt.figure(figsize=(6,6))
        plt.scatter(mus_all[:,0], mus_all[:,1], c=test_labels, s=3, cmap='tab10')
        plt.colorbar(ticks=range(10)); plt.title('VAE latent space (mu) colored by class')
        save_fig(fig, output_dir/"vae_latent_scatter.png")

        # sampling grid
        n = 15
        digit_size = IMG_SIZE
        figure = np.zeros((digit_size * n, digit_size * n))
        grid_x = np.linspace(-3,3,n)
        grid_y = np.linspace(-3,3,n)[::-1]
        with torch.no_grad():
            for i, yi in enumerate(grid_y):
                for j, xi in enumerate(grid_x):
                    z_sample = torch.tensor([[xi, yi]], dtype=torch.float32).to(DEVICE)
                    x_dec = vae.decode(z_sample)
                    x_img = x_dec[0,0].cpu().numpy()
                    figure[i*digit_size:(i+1)*digit_size, j*digit_size:(j+1)*digit_size] = x_img
        fig = plt.figure(figsize=(8,8)); plt.imshow(figure, cmap='gray'); plt.axis('off')
        save_fig(fig, output_dir/"vae_sampling_grid.png")

    # recon samples
    n = 10
    idx = np.random.choice(len(orig), n, replace=False)
    fig = plt.figure(figsize=(20,4))
    for i,k in enumerate(idx):
        ax=plt.subplot(2,n,i+1); plt.imshow(orig[k].squeeze(), cmap='gray'); plt.axis('off')
        ax=plt.subplot(2,n,n+i+1); plt.imshow(recons_all[k].squeeze(), cmap='gray'); plt.axis('off')
    plt.suptitle('VAE: top original, bottom reconstruction', y=1.02)
    save_fig(fig, output_dir/"vae_recon.png")

    # histograms
    fig = plt.figure(figsize=(10,4))
    plt.subplot(1,2,1); plt.hist(psnrs, bins=50); plt.title('PSNR distribution'); plt.xlabel('PSNR')
    plt.subplot(1,2,2); plt.hist(ssims, bins=50); plt.title('SSIM distribution'); plt.xlabel('SSIM')
    save_fig(fig, output_dir/"vae_metrics_hist.png")

    return metrics, mus_all


# ---------------- GAN experiment ----------------
def normalize_minus1_1(x):
    return (x * 2.0) - 1.0


def run_gan(output_dir=OUTPUT_DIR/"GAN", steps=GAN_STEPS, nz=LATENT_GAN, batch=BATCH):
    output_dir = Path(output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)
    train_loader, _, train_dataset, _ = get_dataloaders(batch=batch)

    G = DCGAN_G(nz=nz).to(DEVICE)
    D = DCGAN_D().to(DEVICE)
    g_opt = torch.optim.Adam(G.parameters(), lr=1e-4, betas=(0.5,0.999))
    d_opt = torch.optim.Adam(D.parameters(), lr=1e-4, betas=(0.5,0.999))
    bce_loss = nn.BCEWithLogitsLoss()

    # create loader with floats in [0,1]
    gan_dataset = torch.utils.data.TensorDataset(((train_dataset.data.unsqueeze(1).float()/255.0)), train_dataset.targets)
    gan_loader = DataLoader(gan_dataset, batch_size=batch, shuffle=True, num_workers=2)

    d_losses = []; g_losses = []
    step = 0
    t0 = time.time()
    fixed_noise = torch.randn(25, nz, device=DEVICE)
    while step < steps:
        for xb, _ in gan_loader:
            xb = xb.to(DEVICE)
            xb = normalize_minus1_1(xb)
            bs = xb.size(0)
            D.train(); G.train()
            noise = torch.randn(bs, nz, device=DEVICE)
            fake = G(noise)
            real_logits = D(xb)
            fake_logits = D(fake.detach())
            d_loss = bce_loss(real_logits, torch.ones_like(real_logits)) + bce_loss(fake_logits, torch.zeros_like(fake_logits))
            d_opt.zero_grad(); d_loss.backward(); d_opt.step()
            noise2 = torch.randn(bs, nz, device=DEVICE)
            fake2 = G(noise2)
            fake_logits2 = D(fake2)
            g_loss = bce_loss(fake_logits2, torch.ones_like(fake_logits2))
            g_opt.zero_grad(); g_loss.backward(); g_opt.step()
            d_losses.append(d_loss.item()); g_losses.append(g_loss.item())
            step += 1
            if step % 500 == 0 or step==steps:
                print(f"GAN step {step}/{steps} D_loss {d_loss.item():.4f} G_loss {g_loss.item():.4f}")
                with torch.no_grad():
                    samples = G(fixed_noise).cpu()
                    samples = (samples + 1.0) / 2.0
                    grid = utils.make_grid(samples, nrow=5, normalize=False)
                    plt.figure(figsize=(4,4)); plt.imshow(grid.permute(1,2,0).squeeze(), cmap='gray'); plt.axis('off')
                    save_fig(plt.gcf(), output_dir/f"gan_samples_step_{step}.png")
            if step >= steps:
                break
        if step >= steps:
            break
    gan_time = time.time() - t0
    print("GAN trained steps:", step)

    # final sample grid
    with torch.no_grad():
        samples = G(torch.randn(25, nz, device=DEVICE)).cpu()
        samples01 = (samples + 1.0) / 2.0
        grid = utils.make_grid(samples01, nrow=5, normalize=False)
        plt.figure(figsize=(4,4)); plt.imshow(grid.permute(1,2,0).squeeze(), cmap='gray'); plt.axis('off')
        save_fig(plt.gcf(), output_dir/"gan_final_samples.png")

    # simple diversity metrics on generated samples
    gen = samples01.numpy()  # (N,1,28,28) in [0,1]
    # per-pixel variance across generated images
    pixel_var = float(np.mean(np.var(gen, axis=0)))
    # average pairwise SSIM (small N) to approximate diversity (lower -> more diverse)
    n = gen.shape[0]
    pair_ssim = []
    for i in range(n):
        for j in range(i+1, n):
            pair_ssim.append(ssim(gen[i,0], gen[j,0], data_range=1.0))
    mean_pair_ssim = float(np.mean(pair_ssim)) if len(pair_ssim)>0 else 0.0

    metrics = {"train_time_sec": gan_time, "steps_trained": int(step),
               "pixel_variance": pixel_var, "mean_pairwise_ssim": mean_pair_ssim}
    (output_dir/"metrics.json").write_text(json.dumps(metrics, indent=2))

    # loss curves
    fig = plt.figure(figsize=(6,4))
    plt.plot(d_losses); plt.plot(g_losses); plt.title('GAN D & G loss'); plt.xlabel('Steps'); plt.ylabel('Loss'); plt.legend(['D','G'])
    save_fig(fig, output_dir/"gan_loss.png")

    print("GAN metrics:", metrics)
    return metrics


# ---------------- Main / CLI ----------------
if __name__ == "__main__":
    parser = argparse.ArgumentParser()
    parser.add_argument("--run", choices=["all","ae","vae","gan"], default="all")
    parser.add_argument("--gan_steps", type=int, default=GAN_STEPS)
    parser.add_argument("--epochs_ae", type=int, default=EPOCHS_AE)
    parser.add_argument("--epochs_vae", type=int, default=EPOCHS_VAE)
    # NOTE: In Jupyter/Colab the IPython kernel injects extra args (like -f /root/...).
    # Using parse_known_args() prevents argparse from failing on those unknown flags.
    args, _ = parser.parse_known_args()

    summary = {}
    if args.run in ("all","ae"):
        print("Running AE...")
        summary['ae'] = run_ae(output_dir=OUTPUT_DIR/"AE", epochs=args.epochs_ae)
    if args.run in ("all","vae"):
        print("Running VAE...")
        metrics, mus = run_vae(output_dir=OUTPUT_DIR/"VAE", epochs=args.epochs_vae)
        summary['vae'] = metrics
    if args.run in ("all","gan"):
        print("Running GAN...")
        summary['gan'] = run_gan(output_dir=OUTPUT_DIR/"GAN", steps=args.gan_steps)

    (OUTPUT_DIR/"metrics_summary.json").write_text(json.dumps(summary, indent=2))
    print("Saved metrics summary to", OUTPUT_DIR/"metrics_summary.json")
```
